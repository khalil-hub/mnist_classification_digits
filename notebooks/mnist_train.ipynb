```python
#EDA
import sys
import os
sys.path.append(os.path.abspath('..'))
from tensorflow.keras.datasets import mnist
from src.data_loader import pre_process_data
from tensorflow.keras.utils import to_categorical
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

(x_train, y_train), (x_test, y_test) = mnist.load_data()

print(f"x_train shape: {x_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"x_test shape: {x_test.shape}")
print(f"y_test shape: {y_test.shape}")

print("first 10 labels:", y_train[:10])
print("x: ", x_train[0])
plt.imshow(x_train[0], cmap="gray")
plt.title(f"Label: {y_train[0]}")
plt.axis("off")
plt.show()
x_train, y_train, x_test, y_test = pre_process_data(x_train, y_train, x_test, y_test)
print("shape of y_test: ", y_test.shape)
print("first array: ", y_test[0])
```

```python
#data type
print(x_train.dtype)
print(y_train.dtype)
x_train=x_train.astype('float32')/255.0
x_test=x_test.astype('float32')/255.0
#print(x_train[0])
#Reshaping of images
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)
#print(x_train[0])
#One hot encoding
y_train = to_categorical(y_train, num_classes=10)
y_test=to_categorical(y_test, num_classes=10)
print(y_train[:10])
print(y_train.shape)
```

```python
#check for data unbalance
y_train_series=pd.Series(y_train)
print(y_train_series.describe())
print(y_train_series.value_counts().sort_index())
```

```python
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns

confusion_matrices={}
path='../predictions'
matrix_files = [f for f in os.listdir(path) if f.startswith('confusion')]
#load confusion matrix
for matrix_file in matrix_files:
    matrix_path=os.path.join(path, matrix_file)
    cm=np.load(matrix_path)
    confusion_matrices[matrix_file]=cm
    print(f"matrix_{matrix_file}_loaded")
    plt.figure(figsize=(8, 6))
    sns.heatmap(confusion_matrices[matrix_file], annot=True, fmt='d', cmap='Blues')
    plt.title(f"confusion matrix: {matrix_file}")
    plt.show()

print(confusion_matrices)
cm_best=np.load(os.path.join(path, 'confusion_matrix_mnist_cnn_adam_10.npy'))
correct_predictions_per_class= np.diag(cm_best)
samples_per_class=np.sum(cm_best, axis=1)
per_class_recall=correct_predictions_per_class/samples_per_class
print(per_class_recall)
for i in range(len(per_class_recall)):
    print(f"{i} : {per_class_recall[i]}")
```
